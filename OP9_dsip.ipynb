```python
import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-10-25 231042.png')

kernel = np.ones((5, 5), np.float32) / 25

result = cv2.filter2D(image, -1, kernel)

print("Averaging Filter")
cv2_imshow(result)

cv2.imwrite('/content/image1_averaging.jpg', result)
```

```python

import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-10-25 231042.png')

result = cv2.GaussianBlur(image, (5, 5), 1.5)

print("Gaussian Filter")
cv2_imshow(result)

cv2.imwrite('/content/image1_gaussian.png', result)
```

```python

import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-10-25 231042.png')

result = cv2.medianBlur(image, 5)

print("Median Filter")
cv2_imshow(result)

cv2.imwrite('/content/image1_median.png', result)
```

```python

import cv2
import numpy as np
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-10-25 231042.png')

kernel = np.array([
    [-1, -1, -1],
    [-1,  9, -1],
    [-1, -1, -1]
])

result = cv2.filter2D(image, -1, kernel)

print("High-Pass Sharpened")
cv2_imshow(result)

cv2.imwrite('/content/image1_highpass.png', result)
```

```python

import cv2
from google.colab.patches import cv2_imshow

image = cv2.imread('/content/Screenshot 2026-10-25 231042.png')

laplacian = cv2.Laplacian(image, cv2.CV_64F)
result = cv2.convertScaleAbs(image - laplacian)

print("Laplacian Sharpened")
cv2_imshow(result)

cv2.imwrite('/content/image1_laplacian.png', result)
```
