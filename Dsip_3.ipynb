```python
import numpy as np
import matplotlib.pyplot as plt

x = np.array([1,2,3,2,1])

plt.stem(range(len(x)),x)
plt.title("Discrete-Time signal x[n]")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python
y = np.array([0,1,2,3,2,])

plt.stem(range(len(y)),y)
plt.title("Discrete-Time signal y[n]")
plt.xlabel("Samples")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 2, 1])

auto_corr = np.correlate(x, x, mode='full')

lags = np.arange(-(len(x)-1), len(x))

plt.stem(lags, auto_corr)
plt.title("Autocorrelation of x[n]")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 2, 1])
y = np.array([0, 1, 2, 3, 2])

cross_corr = np.correlate(x, y, mode='full')

lags = np.arange(-(len(x)-1), len(x))

plt.stem(lags, cross_corr)
plt.title("Cross-Correlation of x[n] and y[n]")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)
plt.show()
```

```python

import numpy as np
import matplotlib.pyplot as plt

x = np.array([1, 2, 3, 2, 1])
y = np.array([0, 1, 2, 3, 2])

auto_corr = np.correlate(x, x, mode='full')
cross_corr = np.correlate(x, y, mode='full')

lags = np.arange(-(len(x)-1), len(x))

plt.figure(figsize=(10, 4))

plt.subplot(1, 2, 1)
plt.stem(lags, auto_corr)
plt.title("Autocorrelation")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)

plt.subplot(1, 2, 2)
plt.stem(lags, cross_corr)
plt.title("Cross-Correlation")
plt.xlabel("Lag")
plt.ylabel("Amplitude")
plt.grid(True)

plt.tight_layout()
plt.show()
```
